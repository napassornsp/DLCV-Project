# Exploratory Data Analysis: ISOT Fake and Real News

This notebook explores the combined dataset produced by `combine_dataset.ipynb`. Real articles come from Reuters, and fake articles come from websites flagged as unreliable. It checks data quality and looks for shortcuts that would let a model tell the classes apart without reading the article. Nothing is changed or saved here: the problems it finds are fixed in `clean_dataset.ipynb` and `preprocessing.py`.

## 1. Set up

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display
```

## 2. Load Data

```python
df = pd.read_csv("Dataset/fake_news_combined.csv.gz")
df.head()
```

## 3. Structural Overview

```python
print("Shape:", df.shape)
df.info()
```

## 4. Data Quality Checks

### Missing Values

```python
df.isnull().sum().to_frame("missing_count")
```

No column has missing values in the usual sense. A text can still be present but empty, though, so we also check for blank strings.

```python
for col in ["title", "text"]:
    blank = df[col].str.strip() == ""
    print(f"{col}: {blank.sum()} blank rows")
    print(df.loc[blank, "label"].value_counts().to_string(), "\n")
```

631 articles have an empty text, and 630 of them are fake. These are most likely pages where the scraper captured only a title, for example video posts. An empty text gives a model nothing to read except the title, so these rows are removed during cleaning.

### Extra Spaces

```python
for col in ["title", "text", "date"]:
    padded = df[col] != df[col].str.strip()
    print(f"{col}: {padded.sum()} rows with leading/trailing spaces")
    print(df.loc[padded, "label"].value_counts().to_string(), "\n")
```

Extra spaces follow the label. 9,071 of the 9,135 padded titles are fake, and every padded date (21,417) is real. Each file was scraped differently and left its own formatting behind. This doesn't matter for TF-IDF, which ignores whitespace, but it shows that formatting can give the label away, so all text columns are stripped during cleaning.

### Duplicates

```python
print("Fully duplicated rows:", df.duplicated().sum())

stripped_text = df["text"].str.strip()
dup_text = stripped_text.duplicated(keep=False) & (stripped_text != "")
print("\nRows whose (non-blank) text appears more than once:", dup_text.sum())
print(df.loc[dup_text, "label"].value_counts().to_string())

# Does any article text appear under BOTH labels?
labels_per_text = df[stripped_text != ""].groupby(stripped_text)["label"].nunique()
print("\nTexts that appear with both labels:", (labels_per_text > 1).sum())
```

Only 209 rows are exact copies, but 10,766 rows share their text with at least one other row, and almost all of these (10,327) are fake. The same fake article was often posted under several subjects or with a slightly different title. No text appears under both labels, so the labels never contradict each other.

Duplicates must be removed before the train/test split. Otherwise one copy of an article can end up in training and another in testing, and the model is rewarded for remembering an article rather than judging it.

### Date Parsing

```python
parsed_date = pd.to_datetime(df["date"].str.strip(), format="mixed", errors="coerce")
bad_date = parsed_date.isna()
print("Unparseable dates:", bad_date.sum())
df.loc[bad_date, ["title", "date", "label"]].head(10)
```

10 fake rows have a URL or a headline in the date column, and URLs or "Homepage" in the title. These rows were scraped incorrectly, so they are removed. Every other date parses, even though the two files use different formats ("December 31, 2017" vs "Apr 25, 2017").

## 5. Shortcut Check: Can the Label Be Read Without Reading the Article?

Real and fake articles were collected from different places, so each class may carry fingerprints of where it came from. A model can learn these fingerprints instead of anything about the content, and score very well for the wrong reason.

### Reuters Dateline

```python
DATELINE = r"^\s*.{0,300}?\(Reuters\)\s*[-–—]\s*"

has_dateline = df["text"].str.contains(DATELINE, regex=True)
pd.crosstab(df["label"], has_dateline, normalize="index").mul(100).round(1) \
  .rename(columns={True: "% with dateline", False: "% without"})
```

99.2% of real articles open with a Reuters dateline such as "WASHINGTON (Reuters) - ", sometimes after a correction note or a byline. No fake article does. This one phrase would separate the classes almost perfectly, but it says where an article was published, not whether its content is true, so a model should not be allowed to use it.

### Subject

```python
pd.crosstab(df["subject"], df["label"])
```

The subject column gives away the label completely. Real articles only have `politicsNews` or `worldnews`, and fake articles only have the other six values. The two sources simply used different section names, so `subject` must never be used as a model feature.

### Date

```python
date_check = df.assign(year=parsed_date.dt.year)
pd.crosstab(date_check["year"], date_check["label"])
```

The two classes also cover different periods. Every article from 2015 and 2018 is fake, and real articles are concentrated in 2017. Date is therefore another shortcut and must also stay out of the models. Words tied to a period (names of events in the news at the time) can carry some of the same information, which the explanation methods should watch for.

### Other Source Fingerprints

```python
artefacts = {
    "'Reuters' anywhere in text":       df["text"].str.contains("Reuters"),
    "curly apostrophe (’) in text":     df["text"].str.contains("’"),
    "'image via' in text":              df["text"].str.contains("image via", case=False),
    "URL (http) in text":               df["text"].str.contains("http"),
    "Twitter handle (@) in text":       df["text"].str.contains(r"@\w", regex=True),
    "VIDEO / WATCH in title":           df["title"].str.contains(r"\b(?:VIDEO|WATCH)\b", regex=True),
    "title fully in capitals":          df["title"].str.strip().str.isupper(),
}
artefact_table = pd.DataFrame({name: flag.groupby(df["label"]).mean() for name, flag in artefacts.items()}).T
artefact_table = (artefact_table[["real", "fake"]] * 100).round(1)
artefact_table.columns = ["% of real", "% of fake"]
artefact_table
```

Several fingerprints belong almost entirely to one class:

- **Curly quotes and other non-ASCII characters (’ “ ” – —)** appear only in real articles (47.4% of real articles contain ’). Fake articles contain no non-ASCII characters at all, since their apostrophes were replaced by spaces ("couldn t"). If left in, these symbols would become tokens that only real articles have, so `preprocess_text()` replaces them with spaces.
- **"Reuters"** appears in 99.8% of real articles. Most of these are the dateline, but the name also appears in the body ("told Reuters").
- **"image via"** (photo credits such as "Featured image via Getty Images") appears in 26.7% of fake articles and no real ones.
- **URLs, Twitter handles, and VIDEO/WATCH in titles** are almost exclusive to fake articles.

Some of these, like sensational titles, are plausibly part of what fake news looks like. Others, like photo credits and apostrophe encoding, are side effects of how each website was scraped.

## 6. Target Variable: `label`

```python
print("Counts:")
print(df["label"].value_counts())
print("\nPercentages:")
df["label"].value_counts(normalize=True).mul(100).round(1)
```

```python
ax = sns.countplot(data=df, x="label")
plt.title("Class Balance: Fake vs Real")
for container in ax.containers:
    ax.bar_label(container)
plt.show()
```

52.3% of articles are fake (23,481 vs 21,417 real), so the classes are close to balanced. Almost all of the duplicates and empty texts found above are fake, so without them the share of fake articles would be about 45%. That is still balanced.

## 7. Subject and Date vs Label

```python
sns.countplot(data=df, y="subject", hue="label", order=df["subject"].value_counts().index)
plt.title("Subject vs Label")
plt.show()
```

```python
df["year_month"] = parsed_date.dt.to_period("M")
monthly = df.groupby(["year_month", "label"]).size().unstack(fill_value=0)

monthly.plot(kind="line", figsize=(12, 5))
plt.title("Articles Over Time by Label")
plt.ylabel("Count")
plt.show()
```

The plots confirm Section 5. Each subject bar has only one color, and the two classes peak at different times: real articles rise sharply in late 2017, while fake articles are spread across 2015 to 2017.

## 8. Title and Text Length

```python
df["title_word_count"] = df["title"].str.split().str.len()
df["text_word_count"] = df["text"].str.split().str.len()

display(df.groupby("label")[["title_word_count", "text_word_count"]].describe().T.round(1))

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.boxplot(data=df, x="label", y="title_word_count", ax=axes[0])
axes[0].set_title("Title Length (words) vs Label")
sns.boxplot(data=df, x="label", y="text_word_count", ax=axes[1], showfliers=False)
axes[1].set_title("Text Length (words) vs Label, outliers hidden")
plt.tight_layout()
plt.show()
```

Fake titles are clearly longer: 14 words at the median, against 10 for real titles, and real titles barely vary (standard deviation 1.7). Reuters follows a strict headline style, while fake sites write longer, more sensational headlines.

Text lengths overlap almost completely (medians of 363 and 359 words). The minimum of 0 words is the empty texts found in Section 4.

## 9. Words That Separate the Classes

Which words appear in a much larger share of one class than the other (share of articles containing the word at least once, English stopwords removed).

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(stop_words="english", min_df=20, binary=True)
counts = cv.fit_transform(df["title"] + " " + df["text"])
is_fake = (df["label_encoded"] == 1).to_numpy()

word_share = pd.DataFrame({
    "term": cv.get_feature_names_out(),
    "% of real": np.asarray(counts[~is_fake].mean(axis=0)).ravel() * 100,
    "% of fake": np.asarray(counts[is_fake].mean(axis=0)).ravel() * 100,
})
word_share["difference"] = word_share["% of fake"] - word_share["% of real"]

print("Words much more common in FAKE articles:")
display(word_share.nlargest(15, "difference").round(1))
print("\nWords much more common in REAL articles:")
display(word_share.nsmallest(15, "difference").round(1))
```

The words fall into three groups:

- **Source fingerprints:** reuters is in 99.8% of real articles and 1.3% of fake ones, by far the strongest single word, and washington comes mostly from the dateline too. On the fake side, image, featured, images, getty and com come from photo credits and links. These say where an article came from, not what it claims.
- **Style:** fake articles use conversational words (just, like, know, don, right, fact), while real articles use wire-service reporting language: said (93.7% of real articles), told, minister, government, and weekdays ("said on Tuesday").
- **Format:** video and watch reflect fake sites that embed videos.

The dataset clearly carries signal. The question for the explanation methods is whether the models rely on style and content, or on source fingerprints that would not exist outside this dataset.

## Summary

The combined dataset has 44,898 articles (52.3% fake) and no missing values. Exploration found:

- 631 empty texts (630 fake)
- 10,766 rows that share their text with another row, almost all fake
- 10 badly scraped rows with a URL or headline in the date column
- a Reuters dateline at the start of 99.2% of real articles
- curly quotes and other non-ASCII characters in real articles only (47.4% of real articles contain ’), and none in fake ones
- `subject` and `date`, which each give away the label
- other source fingerprints: "Reuters" in article bodies, photo credits, URLs and Twitter handles

The row problems and the Reuters dateline are fixed in `clean_dataset.ipynb`, and the non-ASCII characters are handled by `preprocess_text()` in `preprocessing.py`. `subject` and `date` are simply never used as model features.
